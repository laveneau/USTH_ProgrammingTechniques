## Exercise 1
Add typing (parameters and return values) to the following code.
*(It is `x**y` by exponentiation-by-squaring, pure Python — no numpy needed.)*

```python
%%writefile code/exercise1.py
def power(x, y):
    result = 1.0
    while True:
        t = y % 2
        y = y // 2
        if t == 1:
            result = result * x
        if y == 0:
            break
        x = x * x
    return result


if __name__ == "__main__":
    print(f"power(2, 10)   = {power(2, 10)}")
    print(f"power(2.0, 10) = {power(2.0, 10)}")
```

## Exercise 3
Add typing to every parameter and returned value of the following linked list
(adapted from [realpython](https://realpython.com/linked-lists-python/)).
*Tip: the nested `__Node` is private and self-referential — a `from __future__ import annotations`
and a type alias help.*

```python
%%writefile code/exercise3.py
from __future__ import annotations


class LinkedList:
    class _Node:
        def __init__(self, data, next=None):
            self.__data = data
            self.__next = next

        @property
        def data(self):
            return self.__data

        @property
        def next(self):
            return self.__next

        @next.setter
        def next(self, _next):
            self.__next = _next

        def __repr__(self):
            return repr(self.data)

    def __init__(self, nodes=None):
        self.__head = None
        if nodes is not None:
            node = self._Node(data=nodes.pop(0))
            self.__head = node
            for elem in nodes:
                node.next = self._Node(elem)
                node = node.next

    def __repr__(self):
        node = self.__head
        nodes = []
        while node is not None:
            nodes.append(repr(node))
            node = node.next
        nodes.append("None")
        return " -> ".join(nodes)

    def __iter__(self):
        node = self.__head
        while node is not None:
            yield node.data
            node = node.next

    def is_empty(self):
        return self.__head is None

    def add_first(self, data):
        self.__head = self._Node(data, self.__head)
        return self

    def add_last(self, data):
        if self.is_empty():
            return self.add_first(data)
        current_node = self.__head
        while current_node.next is not None:
            current_node = current_node.next
        current_node.next = self._Node(data)
        return self


if __name__ == "__main__":
    lklist = LinkedList(["b", "d", "f"])
    lklist.add_first("a")
    lklist.add_last("g")
    print(lklist)
```
